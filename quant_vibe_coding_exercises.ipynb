#### 📈 **Exercise 1: Interactive Stock Price Chart with Plotly**

Create an interactive line chart showing stock price movement over time using Plotly.

**Requirements:**
- Generate synthetic daily stock prices for 252 trading days (1 year)
- Starting price: $100
- Use random daily returns between -3% and +3%
- Create a Plotly line chart with:
  - X-axis: Trading days (0 to 251)
  - Y-axis: Stock price
  - Title: "Stock Price Movement - 1 Year"
  - Hover information showing day and exact price

**💡 Hint:** Import `plotly.graph_objects`, `numpy`, and use cumulative returns to generate prices.


```python
import plotly.graph_objects as go
import numpy as np

# Generate synthetic data
np.random.seed(42)
days = np.arange(252)
daily_returns = np.random.uniform(-0.03, 0.03, 252)
prices = 100 * np.cumprod(1 + daily_returns)

# Create interactive plot
fig = go.Figure()
fig.add_trace(
    go.Scatter(
        x=days,
        y=prices,
        mode='lines',
        name='Stock Price',
        hovertemplate='Day: %{x}<br>Price: $%{y:.2f}'
    )
)

fig.update_layout(
    title='Stock Price Movement - 1 Year',
    xaxis_title='Trading Days',
    yaxis_title='Stock Price ($)',
    hovermode='x unified'
)

fig.show()
```

#### 🕯️ **Exercise 2: Candlestick Chart with Plotly**

Create a candlestick chart to visualize OHLC (Open, High, Low, Close) data.

**Requirements:**
- Generate synthetic OHLC data for 30 trading days
- Starting price: $150
- Each day's data should follow realistic patterns:
  - Open = Previous close ± small random change
  - High = max(Open, Close) + random value
  - Low = min(Open, Close) - random value
  - Close = Open ± random daily change
- Create a Plotly candlestick chart with proper date labels
- Add title: "30-Day Candlestick Chart"

**💡 Hint:** Use `plotly.graph_objects.Candlestick()` and `pandas` for date handling.


```python
import plotly.graph_objects as go
import pandas as pd
import numpy as np

# Generate dates
dates = pd.date_range(start='2023-01-01', periods=30, freq='B')

# Initialize price arrays
opens = np.zeros(30)
highs = np.zeros(30)
lows = np.zeros(30)
closes = np.zeros(30)

# Set initial values
opens[0] = 150
closes[0] = opens[0] * (1 + np.random.uniform(-0.02, 0.02))
highs[0] = max(opens[0], closes[0]) * (1 + np.random.uniform(0, 0.01))
lows[0] = min(opens[0], closes[0]) * (1 - np.random.uniform(0, 0.01))

# Generate OHLC data
for i in range(1, 30):
    opens[i] = closes[i-1] * (1 + np.random.uniform(-0.01, 0.01))
    closes[i] = opens[i] * (1 + np.random.uniform(-0.02, 0.02))
    highs[i] = max(opens[i], closes[i]) * (1 + np.random.uniform(0, 0.01))
    lows[i] = min(opens[i], closes[i]) * (1 - np.random.uniform(0, 0.01))

# Create candlestick chart
fig = go.Figure(data=[go.Candlestick(
    x=dates,
    open=opens,
    high=highs,
    low=lows,
    close=closes
)])

fig.update_layout(
    title='30-Day Candlestick Chart',
    yaxis_title='Price ($)',
    xaxis_title='Date'
)

fig.show()
```

#### 📊 **Exercise 4: Simple Linear Regression on Stock Prices**

Perform a linear regression to predict stock prices and visualize the results.

**Requirements:**
- Generate synthetic stock price data for 100 days with an upward trend
- Use days (0-99) as X variable
- Stock price as Y variable (start at $100, add linear trend + noise)
- Perform linear regression using `scipy.stats.linregress` or `sklearn`
- Create a Plotly scatter plot with:
  - Original data points
  - Regression line overlay
  - Display R² value in the title
  - Title: "Linear Regression: Stock Price vs Time (R²=X.XX)"

**💡 Hint:** Use `scipy.stats.linregress()` or `sklearn.linear_model.LinearRegression()`.


```python
import plotly.graph_objects as go
import numpy as np
from scipy import stats

# Generate synthetic data
np.random.seed(42)
days = np.arange(100)
trend = 0.5 * days  # Add $0.50 per day trend
noise = np.random.normal(0, 5, 100)
prices = 100 + trend + noise

# Perform linear regression
slope, intercept, r_value, p_value, std_err = stats.linregress(days, prices)
r_squared = r_value**2

# Calculate regression line
line = slope * days + intercept

# Create plot
fig = go.Figure()

# Add scatter plot of original data
fig.add_trace(go.Scatter(
    x=days,
    y=prices,
    mode='markers',
    name='Actual Prices'
))

# Add regression line
fig.add_trace(go.Scatter(
    x=days,
    y=line,
    mode='lines',
    name='Regression Line'
))

fig.update_layout(
    title=f'Linear Regression: Stock Price vs Time (R²={r_squared:.2f})',
    xaxis_title='Days',
    yaxis_title='Price ($)'
)

fig.show()
```

#### 🔥 **Exercise 5: Correlation Heatmap for Portfolio Assets**

Create a correlation heatmap for multiple assets using Plotly.

**Requirements:**
- Generate synthetic daily returns for 5 assets (e.g., AAPL, GOOGL, MSFT, TSLA, AMZN)
- Create 252 days of return data for each asset
- Use numpy to generate correlated returns (some positive, some negative correlations)
- Calculate the correlation matrix using pandas
- Create a Plotly heatmap showing correlations
- Include asset names on both axes
- Use a color scale (e.g., RdBu) with annotations showing correlation values
- Title: "Asset Return Correlations"

**💡 Hint:** Use `plotly.graph_objects.Heatmap()` and `pandas.DataFrame.corr()`.


```python
import plotly.graph_objects as go
import numpy as np
import pandas as pd

# Generate correlated returns
np.random.seed(42)
n_days = 252
n_assets = 5

# Create correlation matrix for generation
corr_matrix = np.array([
    [1.0, 0.7, 0.6, 0.3, -0.2],
    [0.7, 1.0, 0.8, 0.4, -0.1],
    [0.6, 0.8, 1.0, 0.5, -0.3],
    [0.3, 0.4, 0.5, 1.0, 0.2],
    [-0.2, -0.1, -0.3, 0.2, 1.0]
])

# Generate correlated random returns
L = np.linalg.cholesky(corr_matrix)
uncorrelated_returns = np.random.normal(0, 0.01, (n_days, n_assets))
returns = uncorrelated_returns @ L.T

# Create DataFrame
assets = ['AAPL', 'GOOGL', 'MSFT', 'TSLA', 'AMZN']
df = pd.DataFrame(returns, columns=assets)

# Calculate correlation matrix
corr = df.corr()

# Create heatmap
fig = go.Figure(data=go.Heatmap(
    z=corr,
    x=assets,
    y=assets,
    colorscale='RdBu',
    zmin=-1,
    zmax=1,
    text=np.round(corr, 2),
    texttemplate='%{text}',
    textfont={"size": 10},
    showscale=True
))

fig.update_layout(
    title='Asset Return Correlations',
    xaxis_title='Asset',
    yaxis_title='Asset'
)

fig.show()
```

#### 🎰 **Exercise 6: Monte Carlo Simulation with Visualization**

Run a Monte Carlo simulation for stock price paths and visualize the results.

**Requirements:**
- Starting price: $100
- Simulate 1,000 different price paths
- Each path: 252 trading days
- Daily return parameters: mean = 0.05%, std = 1.5%
- Plot all 1,000 paths on a single Plotly chart
- Make individual paths semi-transparent (opacity ~0.1)
- Add the mean path as a bold line
- Include percentile bands (5th and 95th percentiles)
- Title: "Monte Carlo Simulation: 1,000 Price Paths"

**💡 Hint:** Use loops or vectorization with numpy, and `plotly.graph_objects.Scatter()`.


```python
import plotly.graph_objects as go
import numpy as np

# Simulation parameters
n_paths = 1000
n_days = 252
start_price = 100
daily_mean = 0.0005  # 0.05%
daily_std = 0.015    # 1.5%

# Generate returns
np.random.seed(42)
returns = np.random.normal(daily_mean, daily_std, (n_paths, n_days))

# Calculate price paths
price_paths = start_price * np.cumprod(1 + returns, axis=1)

# Calculate statistics
mean_path = np.mean(price_paths, axis=0)
percentile_5 = np.percentile(price_paths, 5, axis=0)
percentile_95 = np.percentile(price_paths, 95, axis=0)

# Create plot
fig = go.Figure()

# Add individual paths
for path in price_paths:
    fig.add_trace(go.Scatter(
        y=path,
        mode='lines',
        opacity=0.1,
        line=dict(color='gray'),
        showlegend=False
    ))

# Add mean path
fig.add_trace(go.Scatter(
    y=mean_path,
    mode='lines',
    line=dict(color='blue', width=3),
    name='Mean Path'
))

# Add percentile bands
fig.add_trace(go.Scatter(
    y=percentile_95,
    mode='lines',
    line=dict(color='red', dash='dash'),
    name='95th Percentile'
))

fig.add_trace(go.Scatter(
    y=percentile_5,
    mode='lines',
    line=dict(color='red', dash='dash'),
    name='5th Percentile'
))

fig.update_layout(
    title='Monte Carlo Simulation: 1,000 Price Paths',
    xaxis_title='Trading Days',
    yaxis_title='Stock Price ($)'
)

fig.show()
```

#### 📉 **Exercise 7: Bollinger Bands Calculation and Plotting**

Calculate and visualize Bollinger Bands for a stock price series.

**Requirements:**
- Generate synthetic stock price data for 100 days
- Calculate 20-day Simple Moving Average (SMA)
- Calculate 20-day rolling standard deviation
- Compute upper band: SMA + (2 × std)
- Compute lower band: SMA - (2 × std)
- Create a Plotly chart with:
  - Stock price line
  - Middle band (SMA)
  - Upper and lower bands
  - Shaded area between bands
- Title: "Bollinger Bands (20-day, 2 std)"

**💡 Hint:** Use `pandas.rolling()` for calculations and `plotly.graph_objects.Scatter()` with `fill='tonexty'`.


```python
import plotly.graph_objects as go
import numpy as np
import pandas as pd

# Generate synthetic price data
np.random.seed(42)
returns = np.random.normal(0.001, 0.02, 100)
prices = 100 * np.cumprod(1 + returns)

# Create DataFrame
df = pd.DataFrame(prices, columns=['Price'])

# Calculate Bollinger Bands
window = 20
df['SMA'] = df['Price'].rolling(window=window).mean()
df['STD'] = df['Price'].rolling(window=window).std()
df['Upper'] = df['SMA'] + (df['STD'] * 2)
df['Lower'] = df['SMA'] - (df['STD'] * 2)

# Create plot
fig = go.Figure()

# Add price line
fig.add_trace(go.Scatter(
    y=df['Price'],
    mode='lines',
    name='Price',
    line=dict(color='blue')
))

# Add SMA
fig.add_trace(go.Scatter(
    y=df['SMA'],
    mode='lines',
    name='20-day SMA',
    line=dict(color='orange')
))

# Add Bollinger Bands
fig.add_trace(go.Scatter(
    y=df['Upper'],
    mode='lines',
    name='Upper Band',
    line=dict(color='gray', dash='dash'),
    showlegend=True
))

fig.add_trace(go.Scatter(
    y=df['Lower'],
    mode='lines',
    name='Lower Band',
    line=dict(color='gray', dash='dash'),
    fill='tonexty',
    fillcolor='rgba(200,200,200,0.2)',
    showlegend=True
))

fig.update_layout(
    title='Bollinger Bands (20-day, 2 std)',
    xaxis_title='Days',
    yaxis_title='Price ($)'
)

fig.show()
```

#### 🏗️ **Exercise 8: Generating Realistic OHLCV Data**

Create a function to generate synthetic OHLCV (Open, High, Low, Close, Volume) data.

**Requirements:**
- Create a function `generate_ohlcv(days, start_price, volatility)`
- Generate data following these rules:
  - Open: Previous close ± random percentage
  - Close: Open × (1 + random return from normal distribution)
  - High: max(Open, Close) × (1 + small random %)
  - Low: min(Open, Close) × (1 - small random %)
  - Volume: Random integer between 1M and 10M shares
- Generate 60 days of data starting at $200
- Create a pandas DataFrame with proper datetime index
- Display first 10 rows and basic statistics

**💡 Hint:** Use `pandas.DataFrame()` and `pandas.date_range()` for datetime index.


```python
import pandas as pd
import numpy as np

def generate_ohlcv(days, start_price, volatility):
    """Generate synthetic OHLCV data"""
    dates = pd.date_range(start='2023-01-01', periods=days, freq='B')
    data = {
        'Open': np.zeros(days),
        'High': np.zeros(days),
        'Low': np.zeros(days),
        'Close': np.zeros(days),
        'Volume': np.random.randint(1_000_000, 10_000_000, days)
    }
    
    # Set initial values
    data['Open'][0] = start_price
    data['Close'][0] = start_price * (1 + np.random.normal(0, volatility))
    
    # Calculate High and Low for first day
    high_low_range = 0.002  # 0.2% range for high/low
    data['High'][0] = max(data['Open'][0], data['Close'][0]) * (1 + np.random.uniform(0, high_low_range))
    data['Low'][0] = min(data['Open'][0], data['Close'][0]) * (1 - np.random.uniform(0, high_low_range))
    
    # Generate remaining days
    for i in range(1, days):
        # Open starts at previous close with some variation
        data['Open'][i] = data['Close'][i-1] * (1 + np.random.normal(0, volatility/2))
        
        # Close is calculated from open with random return
        data['Close'][i] = data['Open'][i] * (1 + np.random.normal(0, volatility))
        
        # High and Low are based on Open and Close
        day_high = max(data['Open'][i], data['Close'][i])
        day_low = min(data['Open'][i], data['Close'][i])
        
        data['High'][i] = day_high * (1 + np.random.uniform(0, high_low_range))
        data['Low'][i] = day_low * (1 - np.random.uniform(0, high_low_range))

    # Create DataFrame with datetime index
    df = pd.DataFrame(data, index=dates)
    
    return df

# Generate 60 days of data
df = generate_ohlcv(days=60, start_price=200, volatility=0.02)

# Display first 10 rows and basic statistics
print("\nFirst 10 rows of generated OHLCV data:")
print(df.head(10))
print("\nBasic statistics:")
print(df.describe())
```

#### 🔬 **Exercise 9: Multiple Linear Regression for Factor Analysis**

Build a multiple regression model to analyze stock returns based on market factors.

**Requirements:**
- Generate synthetic data for 200 observations:
  - Market return (independent variable 1)
  - Interest rate changes (independent variable 2)
  - VIX level (independent variable 3)
  - Stock return (dependent variable)
- Create a relationship: Stock Return = 0.01 + 1.2×Market + 0.5×Interest - 0.3×VIX + noise
- Perform multiple linear regression using sklearn
- Print coefficients, R² score, and intercept
- Create a scatter plot: Predicted vs Actual returns
- Add a diagonal line showing perfect predictions

**💡 Hint:** Use `sklearn.linear_model.LinearRegression()` and fit with multiple features.


```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
import plotly.graph_objects as go

# Generate synthetic data
np.random.seed(42)
n_obs = 200

# Generate independent variables
market_return = np.random.normal(0.001, 0.02, n_obs)  # Market returns
interest_changes = np.random.normal(0, 0.001, n_obs)   # Interest rate changes  
vix_level = np.random.normal(20, 5, n_obs)            # VIX levels

# Create true relationship with noise
noise = np.random.normal(0, 0.01, n_obs)
stock_return = 0.01 + 1.2*market_return + 0.5*interest_changes - 0.3*vix_level + noise

# Combine features into X matrix
X = np.column_stack([market_return, interest_changes, vix_level])

# Create and fit linear regression model
model = LinearRegression()
model.fit(X, stock_return)

# Make predictions
predictions = model.predict(X)

# Print results
print("\nRegression Results:")
print(f"Intercept: {model.intercept_:.4f}")
print("\nCoefficients:")
print(f"Market Beta: {model.coef_[0]:.4f}")
print(f"Interest Rate Sensitivity: {model.coef_[1]:.4f}")
print(f"VIX Sensitivity: {model.coef_[2]:.4f}")
print(f"\nR² Score: {model.score(X, stock_return):.4f}")

# Create scatter plot of predicted vs actual returns
fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=stock_return,
        y=predictions,
        mode='markers',
        name='Predictions',
        marker=dict(color='blue', size=8, opacity=0.6)
    )
)

# Add diagonal line for perfect predictions
min_val = min(stock_return.min(), predictions.min())
max_val = max(stock_return.max(), predictions.max())
fig.add_trace(
    go.Scatter(
        x=[min_val, max_val],
        y=[min_val, max_val],
        mode='lines',
        name='Perfect Prediction',
        line=dict(color='red', dash='dash')
    )
)

fig.update_layout(
    title='Predicted vs Actual Stock Returns',
    xaxis_title='Actual Returns',
    yaxis_title='Predicted Returns',
    showlegend=True
)

fig.show()
```

#### 🎛️ **Exercise 10: Interactive Multi-Asset Dashboard**

Create an interactive dashboard with subplots showing multiple visualizations.

**Requirements:**
- Generate synthetic data for 3 different assets over 252 days
- Create a Plotly figure with subplots (3 rows, 1 column):
  - **Subplot 1:** Price comparison (3 line charts on same plot)
  - **Subplot 2:** Daily returns as bar chart for Asset 1
  - **Subplot 3:** Cumulative returns comparison (3 lines)
- Use different colors for each asset
- Add proper titles to each subplot
- Include a main title: "Multi-Asset Performance Dashboard"
- Add legend and ensure all plots are interactive

**💡 Hint:** Use `plotly.subplots.make_subplots()` and add traces to specific rows/cols.


```python
import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# Generate synthetic data
np.random.seed(42)
days = 252
t = np.linspace(0, days-1, days)

# Generate prices with different trends and volatilities
asset1 = 100 * np.exp(0.0002*t + 0.0001*np.random.randn(days).cumsum())
asset2 = 100 * np.exp(0.0001*t + 0.00015*np.random.randn(days).cumsum())
asset3 = 100 * np.exp(0.0003*t + 0.00012*np.random.randn(days).cumsum())

# Calculate returns
returns1 = np.diff(asset1) / asset1[:-1]
returns2 = np.diff(asset2) / asset2[:-1]
returns3 = np.diff(asset3) / asset3[:-1]

# Calculate cumulative returns
cum_ret1 = (1 + returns1).cumprod() - 1
cum_ret2 = (1 + returns2).cumprod() - 1
cum_ret3 = (1 + returns3).cumprod() - 1

# Create figure with subplots
fig = make_subplots(
    rows=3, cols=1,
    subplot_titles=('Price Comparison', 'Daily Returns - Asset 1', 'Cumulative Returns Comparison'),
    vertical_spacing=0.1
)

# Add price comparison traces
fig.add_trace(go.Scatter(x=t, y=asset1, name='Asset 1', line=dict(color='#636efa')), row=1, col=1)
fig.add_trace(go.Scatter(x=t, y=asset2, name='Asset 2', line=dict(color='#EF553B')), row=1, col=1)
fig.add_trace(go.Scatter(x=t, y=asset3, name='Asset 3', line=dict(color='#00cc96')), row=1, col=1)

# Add daily returns bar chart for Asset 1
fig.add_trace(go.Bar(x=t[1:], y=returns1, name='Asset 1 Returns', marker_color='#636efa'), row=2, col=1)

# Add cumulative returns comparison
fig.add_trace(go.Scatter(x=t[1:], y=cum_ret1, name='Asset 1 Cum Ret', line=dict(color='#636efa')), row=3, col=1)
fig.add_trace(go.Scatter(x=t[1:], y=cum_ret2, name='Asset 2 Cum Ret', line=dict(color='#EF553B')), row=3, col=1)
fig.add_trace(go.Scatter(x=t[1:], y=cum_ret3, name='Asset 3 Cum Ret', line=dict(color='#00cc96')), row=3, col=1)

# Update layout
fig.update_layout(
    height=900,
    title_text="Multi-Asset Performance Dashboard",
    showlegend=True,
    template='plotly_white'
)

# Show figure
fig.show()
```
