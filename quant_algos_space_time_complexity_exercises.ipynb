#### 📊 **Exercise 1: Portfolio Value Calculation**

Analyze the complexity of this portfolio value calculation:

```python
def calculate_portfolio_value(positions):
    total_value = 0
    for position in positions:
        total_value += position['shares'] * position['price']
    return total_value
```

**Questions:**
1. What is the **time complexity**? Why?
2. What is the **space complexity**? Why?
3. How does performance change if we double the number of positions?


# 1. This is Linear time as the function iterates through each position exactly once. Each operation inside the loop is constant. Total time scales linearly with number of positions i.e if we double the input, we double the runtime.

# 2. This is a constant space as the algorithm uses a fixed amount of memory regardless of input size. No additional data structures are created. The memory usage doesn't grow with the input.

# 3. Execturion time will double, memory stays the same. Linear scaling makes it very efficient for large portfolios.

#### 💰 **Exercise 3: Risk Calculation**

Analyze the complexity of this risk calculation:

```python
def calculate_portfolio_risk(weights, covariance_matrix):
    n = len(weights)
    risk = 0
    
    for i in range(n):
        for j in range(n):
            risk += weights[i] * weights[j] * covariance_matrix[i][j]
    
    return risk ** 0.5
```

**Questions:**
1. What is the **time complexity**? Why?
2. What is the **space complexity**? Why?
3. Is this efficient for real-time trading?


# 1. This is a quadratic time complexity. Nested loops iterate through n x n matrix. Each mutiplicated and addition is a constant time complextity. Total time grows quadratically with number of assets.

# 2. This is a constant space complexity becauses only uses a single risk variable. Input arrays not counted in space complexity. Constant extra space regardless of input size.

# 3. May be too slow for high-frequency trading with many assets. Quadratic time complexity becomes significant with large portfolios. Consider optimisations or matrix operations for better performance.

#### 🔄 **Exercise 4: Algorithm Comparison**

Compare these two approaches for finding the maximum return:

**Approach A:**
```python
def find_max_return_v1(returns):
    return max(returns)
```

**Approach B:**
```python
def find_max_return_v2(returns):
    max_return = returns[0]
    for return_val in returns[1:]:
        if return_val > max_return:
            max_return = return_val
    return max_return
```

**Questions:**
1. Which approach has better **time complexity**? Why?
2. Which approach has better **space complexity**? Why?
3. Which would you choose for a high-frequency trading system?


# 1. Both approaches have linear time complexity. Both need to scan through entire array once. Python's max() fuction also uses linear scan internally

# 2. Both use constant extra space. Only store max_return variable. No additional data structures needed.

# 3. Approach A is preffered as it has a built in function optimised in C. More concise, maintainable code. Same complexity but faster in practice.
