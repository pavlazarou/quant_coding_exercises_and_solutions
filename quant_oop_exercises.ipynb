#### 📊 **Exercise 1: Basic Stock Class**

Create a `Stock` class with the following:
- **Attributes**: `symbol`, `price`, `shares`
- **Methods**: `calculate_value()`, `update_price(new_price)`, `display_info()`

Create a stock object for **AAPL** with:
- Symbol: **"AAPL"**
- Price: **$175.50**
- Shares: **100**

**Display the stock information and calculate its total value.**


```python
class Stock:
    def __init__(self, symbol, price, shares):
        self.symbol = symbol
        self.price = price
        self.shares = shares
    
    # Instance method - calculate position value
    def calculate_value(self):
        return self.price * self.shares
    
    # Instance method - updates stock price
    def update_price(self, new_price):
        self.price = new_price

    # Instance method - display stock information
    def display_info(self):
        print(f"Stock: {self.symbol}")
        print(f"Price: ${self.price:.2f}")
        print(f"Shares: {self.shares}")
        print(f"Total Value: ${self.calculate_value():.2f}")

# create a stock object for aapl
aapl = Stock("AAPL", 175.50, 100)

aapl.display_info()
```

#### 💼 **Exercise 2: Portfolio Management**

Create a `Portfolio` class that can:
- **Store multiple stock positions**
- **Calculate total portfolio value**
- **Add new positions**
- **Display portfolio summary**

Create a portfolio with these positions:
- **150 shares** of AAPL at **$175.50**
- **50 shares** of GOOGL at **$2,850.75**
- **200 shares** of MSFT at **$415.25**

**Calculate and display the total portfolio value.**



```python
class Portfolio:
    def __init__(self):
        self.stocks = []
        
    def add_stock(self, stock):
        self.stocks.append(stock)
        
    def total_value(self):
        return sum(stock.calculate_value() for stock in self.stocks)
    
    def display_summary(self):
        print("Portfolio Summary:")
        for stock in self.stocks:
            stock.display_info()
            print()
        print(f"Total Portfolio Value: ${self.total_value():.2f}")

# Create portfolio and add stocks
portfolio = Portfolio()
portfolio.add_stock(Stock("AAPL", 175.50, 150))
portfolio.add_stock(Stock("GOOGL", 2850.75, 50))
portfolio.add_stock(Stock("MSFT", 415.25, 200))

# Display portfolio summary
portfolio.display_summary()
```

#### 📈 **Exercise 3: Trading Order Class**

Create a `TradingOrder` class with:
- **Attributes**: `order_id`, `symbol`, `quantity`, `price`, `order_type`, `status`
- **Methods**: `execute_order()`, `cancel_order()`, `display_order()`

Create three different orders:
1. **Market order** for 100 shares of AAPL at $175.50
2. **Limit order** for 50 shares of GOOGL at $2,800.00
3. **Stop order** for 200 shares of MSFT at $400.00

**Display all orders and execute the market order.**


```python
class TradingOrder:
    def __init__(self, order_id, symbol, quantity, price, order_type):
        self.order_id = order_id
        self.symbol = symbol
        self.quantity = quantity
        self.price = price
        self.order_type = order_type
        self.status = "PENDING"
        
    def execute_order(self):
        self.status = "EXECUTED"
        print(f"Order {self.order_id} executed")
        
    def cancel_order(self):
        self.status = "CANCELLED"
        print(f"Order {self.order_id} cancelled")
        
    def display_order(self):
        print(f"Order {self.order_id}: {self.order_type} order for {self.quantity} {self.symbol} @ ${self.price:.2f}")
        print(f"Status: {self.status}")

# Create orders
market_order = TradingOrder(1, "AAPL", 100, 175.50, "MARKET")
limit_order = TradingOrder(2, "GOOGL", 50, 2800.00, "LIMIT") 
stop_order = TradingOrder(3, "MSFT", 200, 400.00, "STOP")

# Display all orders
for order in [market_order, limit_order, stop_order]:
    order.display_order()
    print()
    
# Execute market order
market_order.execute_order()
```

#### 🏦 **Exercise 4: Bank Account Class**

Create a `BankAccount` class with:
- **Attributes**: `account_number`, `balance`, `account_type`
- **Methods**: `deposit(amount)`, `withdraw(amount)`, `check_balance()`, `transfer(other_account, amount)`

Create two accounts:
- **Checking account** with $10,000 initial balance
- **Savings account** with $25,000 initial balance

**Transfer $2,000 from savings to checking and display both balances.**


```python
class BankAccount:
    def __init__(self, account_number, balance, account_type):
        self.account_number = account_number
        self.balance = balance
        self.account_type = account_type
    
    def deposit(self, amount):
        self.balance += amount
        print(f"Deposited ${amount:.2f}")
    
    def withdraw(self,amount):
        if self.balance >= amount:
            self.balance -= amount
            print(f"Withdrew ${amount:.2f}")
        else:
            print("Insufficent Funds")
    
    def check_balance(self):
        print(f"{self.account_type} Balance: ${self.balance:.2f}")
    
    def transfer(self, other_account, amount):
        if self.balance >= amount:
            self.withdraw(amount)
            other_account.deposit(amount)
            print(f"Transferred ${amount:.2f} to account {other_account.account_number}")
        else: 
            print("Insufficent funds for transfer")
        
# create accounts
checking = BankAccount("CHK001", 10_000, "Checking")
savings = BankAccount("SAV001", 25_000, "Savings")

# Transfer and display balances
savings.transfer(checking, 2000)
checking.check_balance()
savings.check_balance()
```

#### 📊 **Exercise 5: Financial Instrument Inheritance**

Create a parent class `FinancialInstrument` with:
- **Attributes**: `symbol`, `price`
- **Methods**: `get_value()`, `get_risk_level()`

Create child classes:
- **`Stock`**: inherits from FinancialInstrument, adds `shares` attribute
- **`Bond`**: inherits from FinancialInstrument, adds `face_value` and `coupon_rate` attributes

Create objects:
- **Stock**: AAPL, $175.50, 100 shares
- **Bond**: T-BOND, $950, $1000 face value, 3% coupon

**Display both instruments with their values and risk levels.**


```python
class FinancialInstrument:
    def __init__(self, symbol, price):
        self.symbol = symbol
        self.price = price

    def get_value(self):
        return self.price
    
    def get_risk_level(self):
        return "Medium"

class Stock(FinancialInstrument):
    def __init__(self, symbol, price, shares):
        super().__init__(symbol,price)
        self.shares = shares

    def get_value(self):
        return self.price * self.shares
    
    def get_risk_level(self):
        return "HIGH"
    
class Bond(FinancialInstrument):
    def __init__(self, symbol, price, face_value, coupon_rate):
        super().__init__(symbol, price)
        self.face_value = face_value
        self.coupon_rate = coupon_rate

    def get_value(self):
        return self.face_value * self.coupon_rate + self.price
    
    def get_risk_level(self):
        return "LOW"

# Create and display instruments
stock = Stock("AAPL", 175.50, 100)
bond = Bond("T-BOND", 950, 1000, 0.03)

print(f"{stock.symbol}: Value = ${stock.get_value():.2f}, Risk = {stock.get_risk_level()}")
print(f"{bond.symbol}: Value = ${bond.get_value():.2f}, Risk = {bond.get_risk_level()}")
                       
```

#### 🎯 **Exercise 6: Trading Strategy Polymorphism**

Each strategy should have different implementations of `analyze_market()` and `execute_signal()`.

Create a function `run_strategies(strategies)` that works with any strategy type (polymorphism).

**Test both strategies on the same market data.**



```python
class TradingStrategy:
    def analyse_market(self):
        pass

    def execute_signal(self):
        pass

class MomentumStrategy(TradingStrategy):
    def analyse_market(self):
        return "Upward price momentum detected"
    
    def execute_signal(self):
        return "BUY based on momentum"
    
class ValueStrategy(TradingStrategy):
    def analyse_market(self):
        return "Stock appears undervalued"
    
    def execute_signal(self):
        return "BUY based on value"

def run_strategies(strategies):
    for strategy in strategies:
        analysis = strategy.analyse_market()
        signal = strategy.execute_signal()
        print(f"Analysis: {analysis}")
        print(f"Signal: {signal}\n")

# Test strategies
strategies = [MomentumStrategy(), ValueStrategy()]
run_strategies(strategies)
```

#### 💰 **Exercise 7: Portfolio with Cash Management**

Enhance the `Portfolio` class to include:
- **Cash balance tracking**
- **Buy shares** method (reduces cash, adds position)
- **Sell shares** method (increases cash, reduces position)
- **Available cash** calculation

Create a portfolio with **$100,000** initial cash and:
1. **Buy** 100 shares of AAPL at $175.50
2. **Buy** 50 shares of GOOGL at $2,850.75
3. **Sell** 25 shares of AAPL at $180.00

**Display the final portfolio value and remaining cash.**



```python
class Portfolio:
    def __init__(self,initial_cash):
        self.cash = initial_cash
        self.positions = {}

    def buy_shares(self, symbol, shares, price):
        cost = shares * price
        if cost <= self.cash:
            self.cash -= cost
            self.positions[symbol] = self.positions.get(symbol, 0) + shares
            return True
        return False
    
    def sell_shares(self, symbol, shares, price):
        if symbol in self.positions and self.positions[symbol] >= shares:
            self.cash += shares * price
            self.positions[symbol] -= shares
            return True
        return False
    
    def get_value(self, prices):
        return self.cash + sum(self.positions[symbol] * prices[symbol] for symbol in self.positions)

# Test Portfolio

portfolio = Portfolio(100_000)
portfolio.buy_shares("AAPL", 100, 175.50)
portfolio.buy_shares("GOOGL", 50, 2850.75)
portfolio.sell_shares("AAPL", 25, 180.00)

current_prices = {"AAPL": 180.00, "GOOGL": 2850.75}
print(f"Portfolio value: ${portfolio.get_value(current_prices):,.2f}")
print(f"Remaining Cash: ${portfolio.cash:,.2f}")
```

#### 📊 **Exercise 8: Risk Management Class**

Create a `RiskManager` class that can:
- **Check position limits** (max 10% of portfolio per stock)
- **Validate trade sizes** (minimum $1,000, maximum $50,000)
- **Calculate portfolio risk metrics** (total exposure, diversification)

Create a portfolio with **$500,000** and test these scenarios:
1. **Valid trade**: $25,000 position in AAPL
2. **Invalid trade**: $75,000 position in GOOGL (exceeds limit)
3. **Invalid trade**: $500 position in MSFT (too small)

**Display risk analysis for each trade attempt.**



```python
class RiskManager:
    def __init__(self, portfolio):
        self.portfolio = portfolio
        self.min_trade = 1000
        self.max_trade = 50_000
        self.position_limit = 0.1

    def validate_trade_size(self,symbol, shares, price):
        trade_value = shares * price
        portfolio_value = self.portfolio.get_value({"AAPL": 175.50, "GOOGL": 2850.75, "MSFT": 415.25})

        if trade_value < self.min_trade:
            print(f"Trade rejected: Below minimum value of ${self.min_trade:,.2f}")
            return False
        if trade_value > self.max_trade:
            print(f"Trade rejected: Above maximum value of ${self.max_trade:,.2f}")
            return False
        if trade_value > portfolio_value * self.position_limit:
            print(f"Trade rejected: Exceeds position limit of {self.position_limit:.0%}")
        
        print(f"Trade validated: ${trade_value:,.2f}")
        return True

# Test scenarios
portfolio = Portfolio(500_000)
risk_mgr = RiskManager(portfolio)

print("\nScenario 1: $25,000 AAPL trade")
risk_mgr.validate_trade_size("AAPL", 142, 175.50)

print("\nScenario 2: $75,000 GOOGL trade")
risk_mgr.validate_trade_size("GOOGL", 26, 2850.75)

print("\nScenario 3: $500 MSFT trade")
risk_mgr.validate_trade_size("MSFT", 1, 415.25)
```

#### 🔄 **Exercise 9: Market Data Class with Updates**

Create a `MarketData` class that can:
- **Store price history** for multiple stocks
- **Update prices** with timestamps
- **Calculate price changes** and percentage returns
- **Find best and worst performers**

Simulate price updates for these stocks:
- **AAPL**: $175.50 → $180.25 → $178.75
- **GOOGL**: $2,850.75 → $2,900.00 → $2,875.50
- **MSFT**: $415.25 → $420.00 → $418.50

**Display price changes and identify the best performer.**



```python
class MarketData:
    def __init__(self):
        self.prices = {}
    
    def update_price(self, symbol, price, timestamp):
        if symbol not in self.prices:
            self.prices[symbol] = []
        self.prices[symbol].append((timestamp,price))
    
    def get_returns(self, symbol):
        if symbol not in self.prices or len(self.prices[symbol]) < 2:
            return 0
        initial = self.prices[symbol][0][1]
        final = self.prices[symbol][-1][1]
        return (final - initial) / initial * 100
    
    def get_best_performer(self):
        returns = [(s, self.get_returns(s)) for s in self.prices]
        return max(returns, key = lambda x: x[1])[0]
    
# test the implementation
market = MarketData()

# add price updates
for ts, aapl in enumerate([175.50, 180.25, 178.75]):
    market.update_price("AAPL", aapl, ts)

for ts, googl in enumerate([2850.75, 2900.00, 2875.00]):
    market.update_price("GOOGL", googl, ts)

for ts, msft in enumerate([415.25, 420.00, 418.50]):
    market.update_price("MSFT", msft, ts)

# Display results
for symbol in market.prices:
    print(f"{symbol} return: {market.get_returns(symbol):.2f}%")
print(f"\nBest performer: {market.get_best_performer()}")
```

#### 🏆 **Exercise 10: Complete Trading System**

Create a comprehensive trading system by combining all previous concepts:

**Components needed:**
1. **`Stock`** class (from Exercise 1)
2. **`Portfolio`** class with cash management (from Exercise 7)
3. **`TradingOrder`** class (from Exercise 3)
4. **`RiskManager`** class (from Exercise 8)
5. **`MarketData`** class (from Exercise 9)

**Scenario:**
- Create a portfolio with **$1,000,000**
- Track AAPL, GOOGL, and MSFT prices
- Place orders: 5 AAPL shares, 1 GOOGL shares, 1000 MSFT shares
- Validate all trades through risk management
- Print the subsequent cash position of the portfolio and stock positions

**Display a complete trading report with all positions and performance.**



```python
market = MarketData()
portfolio = Portfolio(1_000_000)
risk_mgr = RiskManager(portfolio)

# Initialise stock prices
initial_prices = {
    "AAPL" : 175.50,
    "GOOGL" : 2850.75,
    "MSFT" : 415.25
}

for symbol, price in initial_prices.items():
    market.update_price(symbol, price, 0)

# Create and validate order
orders = [
    TradingOrder(1,"AAPL", 5, initial_prices["AAPL"], "BUY"),
    TradingOrder(2,"GOOGL", 1, initial_prices["GOOGL"], "BUY"),
    TradingOrder(3,"MSFT", 1000, initial_prices["MSFT"], "BUY")
]

for order in orders:
    if risk_mgr.validate_trade_size(order.symbol, order.quantity, order.price):
        portfolio.buy_shares(order.symbol, order.quantity, order.price)
        order.execute_order()
    else:
        order.cancel_order()
        
# Generate and print concise trading report
print("\n=== Trading Report ===")
print(f"Cash Position: ${portfolio.cash:,.2f}")
print("\nStock Positions:")
for symbol, shares in portfolio.positions.items():
    print(f"{symbol}: {shares} shares")
```
