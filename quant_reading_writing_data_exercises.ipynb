#### 📖 **Exercise 1: Text File Processing**

Create a function to read market data from a CSV file and calculate the average price:

```python
def analyze_market_data(filename):
    # Your implementation here
    pass
```

**Requirements:**
1. Read CSV file with columns: symbol, price, volume
2. Calculate average price for each symbol
3. Handle file not found errors
4. Return results as a dictionary


````python
filename = 'market_data.csv'

def analyse_market_data(filename):
    try:
        # initialising dictionary to store results
        average_prices = {}
        symbol_totals = {}
        symbol_counts = {}

        # process and read the csv file
        with open(filename, 'r') as file:
            # skip header line
            header = file.readline()

            # process each line
            for line in file:
                # parse csv file
                symbol, price, volume = line.strip().split(',')
                price = float(price)

                # Update running totals
                if symbol not in symbol_totals:
                    symbol_totals[symbol] = price
                    symbol_counts[symbol] = 1
                else:
                    symbol_totals[symbol] += price
                    symbol_counts[symbol] += 1

        # Calculate averages
        for symbol in symbol_totals:
            average_prices[symbol] = symbol_totals[symbol] / symbol_counts[symbol]

        return average_prices
    
    except FileNotFoundError:
        print(f"Error Filename '{filename}' not found")
        return {}
    except Exception as e:
        print(f"Error processing file: {str(e)}")
        return {}
    
# Example usage:
result = analyse_market_data(filename)
print(result)
````

#### ✍️ **Exercise 2: Report Generation**

Create a function to generate a trading performance report:

```python
def generate_performance_report(trades, filename):
    # Your implementation here
    pass
```

**Requirements:**
1. Calculate total profit/loss
2. Count number of trades
3. Write formatted report to file
4. Include summary statistics


````python
filename = 'performance_report.txt'

def generate_performance_report(trades, filename):
    try:
        total_profit = 0
        num_trades = len(trades)

        for trade in trades:
            profit = trade['quantity'] * trade['price']
            total_profit += profit

        # Generate report content
        report = f"Trading Performance Report\n"
        report += f"========================\n\n"
        report += f"Total Number of Trades: {num_trades}\n"
        report += f"Total Profit/Loss: ${total_profit:,.2f}\n"
        report += f"Average Profit per Trade: ${total_profit/num_trades:,.2f}\n\n"
        
        report += "Trade Details:\n"
        report += "-------------\n"
        for i, trade in enumerate(trades, 1):
            report += f"Trade {i}:\n"
            report += f"  Symbol: {trade['symbol']}\n"
            report += f"  Price: ${trade['price']:,.2f}\n"
            report += f"  Quantity: {trade['quantity']}\n"
            report += f"  Trade Value: ${trade['price'] * trade['quantity']:,.2f}\n\n"
            
        # Write report to file
        with open(filename, 'w') as f:
            f.write(report)
            
        print(f"Report successfully written to {filename}")
        
    except Exception as e:
        print(f"Error generating report: {str(e)}")

````

#### 🥒 **Exercise 3: Pickle Serialization**

Create a function to save and load portfolio data:

```python
def save_portfolio(portfolio, filename):
    # Your implementation here
    pass

def load_portfolio(filename):
    # Your implementation here
    pass
```

**Requirements:**
1. Save portfolio as pickle file
2. Load portfolio from pickle file
3. Handle serialization errors
4. Validate loaded data


````python
import pickle

filename = 'portfolio.pkl'

def save_portfolio(portfolio, filename):
    try:
        # Validate Portfolio data
        if not isinstance(portfolio, dict):
            raise ValueError("Portfolio must be a dictionary")
        
        # Save to pickle file
        with open(filename, 'wb') as file:
            pickle.dump(portfolio, file)
        print(f"Portfolio saved to {filename}")

    except Exception as e:
        print(f"Error saving file: {str(e)}")

def load_portfolio(filename):
    try:
        # Load pickle data
        with open(filename, 'rb') as file:
            portfolio = pickle.load(file)
        
        # Validate data
        if not isinstance(portfolio, dict):
            raise ValueError("Loaded data is not a valid portfolio")
        
        print(f"Loaded portfolio successfully from {filename}")
        return portfolio
    
    except Exception as e:
        print(f"Error loading portfolio: {str(e)}")
        return None
        
````
