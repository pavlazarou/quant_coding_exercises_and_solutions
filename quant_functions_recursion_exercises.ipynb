#### 📊 **Exercise 1: Portfolio Value Function**

Create a function that calculates the total value of a portfolio given a list of positions.

**Function should:**
- Take a list of tuples: `[(shares, price), (shares, price), ...]`
- Return the total portfolio value
- Use the function to calculate value for: `[(100, 150.25), (50, 175.50), (200, 85.75)]`


```python
portfolio = [(100, 150.25), (50, 175.50), (200, 85.75)]

# Using a Map with lambda
portfolio_value = list(map(lambda r: r[0] * r[1], portfolio))

total_port_value = sum(portfolio_value)

print(f"Total Portfolio Value (Map with lambda): ${total_port_value:.2f}")


#   Using Loop
def calculate_portfolio_value(positions):
    total_value = 0
    for shares, price in positions:
        total_value += shares * price
    return total_value
                
print(f"Total Portfolio Value (Loop): ${calculate_portfolio_value(portfolio):.2f}")
```

#### 📈 **Exercise 2: Multiple Return Values**

Create a function that analyzes a stock position and returns multiple metrics.

**Function should:**
- Take: `entry_price`, `current_price`, `shares`
- Return: `profit_loss`, `percentage_return`, `current_value`
- Test with: entry=100, current=110, shares=50


```python
def stock_analysis(entry_price, current_price, shares):
    profit_loss = current_price * shares - entry_price * shares
    percentage_return = ((current_price - entry_price) / entry_price) * 100
    current_value = current_price * shares
    return profit_loss, percentage_return, current_value

pl, pct_return, value = stock_analysis(100, 110, 50)

print(f"PnL: ${pl:.2f}")
print(f"Percent Return: {pct_return:.2f}%")
print(f"Current Value: ${value:.2f}")

```

#### ⚡ **Exercise 3: Lambda Functions**

Use lambda functions to process financial data.

**Tasks:**
1. Create a lambda to calculate 5% commission: `commission = lambda price: ...`
2. Apply it to prices: `[100, 150, 200, 175]`
3. Create a lambda to filter positive returns from: `[0.05, -0.02, 0.08, -0.01, 0.03]`



```python
prices = [100, 150, 200, 175]
# Calculate Commission
commission = list(map(lambda price: price * 0.05, prices))
print(f"Commissions: {commission}")

# Calculate Positive Returns
returns = [0.05, -0.02, 0.08, -0.01, 0.03]
positive_returns = list(filter(lambda p_ret: p_ret > 0, returns))
print(f"Positive Returns: {positive_returns}")
```

#### 🗺️ **Exercise 4: Map Function**

Use the `map()` function to transform financial data.

**Tasks:**
1. Convert string prices to floats: `["100.50", "150.25", "200.75"]`
2. Calculate 10% price increases for: `[100, 150, 200, 175]`
3. Calculate position values: shares `[10, 20, 30]` × prices `[100, 150, 200]`


```python
prices = ["100.50", "150.25", "200.75"]

# Convert string prices to floats
float_prices = list(map(float, prices))
print(f"Float Prices: {float_prices}")

# Calculate 10% price increases
initial_prices = [100, 150, 200, 175]
inc_prices = list(map(lambda p: p * 1.10, initial_prices))
print(f"Increased Prices: {inc_prices}")

# Calculate position values
shares = [10, 20, 30]
prices = [100, 150, 200]
position_values = list(map(lambda x, y: x * y, shares, prices))

print(f"Position Values: {position_values}")
```

#### 🔁 **Exercise 5: Recursion**

Create a recursive function to calculate compound interest.

**Function should:**
- Take: `principal`, `rate`, `years`
- Return: final amount after compound interest
- Use recursion: `compound_interest(principal * (1 + rate), rate, years - 1)`
- Test with: principal=1000, rate=0.05, years=3


```python
def calculate_compound_interest(principal, rate, years):
    # Base Case: when years = 0, return principal
    if years == 0:
        return principal
    else:
        return calculate_compound_interest(principal * (1 + rate), rate, years - 1 )

# Test the function
principal = 1000
rate = 0.05
years = 3

result = calculate_compound_interest(principal, rate, years)
print(f"Final amount after {years} years: ${result:.2f}")
```

#### 🎯 **Exercise 6: Fibonacci Trading Levels**

Use recursion to calculate Fibonacci retracement levels for trading.

**Function should:**
- Take: `high_price`, `low_price`, `level` (0-1, where 1 is full retracement)
- Return: price at that Fibonacci level
- Formula: `high - (high - low) * level`
- Use recursion to calculate multiple levels: 0.236, 0.382, 0.618


```python
def fibonacci_retracement(high_price, low_price, levels, index = 0):
    # base case: when we're processing all levels
    if index >= len(levels):
        return []
    
    # Calculate current level
    current_level = high_price - (high_price - low_price) * levels[index]

    # Recursive case: calculate current level and continue with next level
    return [current_level] + fibonacci_retracement(high_price, low_price, levels, index + 1)

# Test the function
high = 100
low = 50
fib_levels = [0.236, 0.382, 0.618]

retracement_levels = fibonacci_retracement(high, low, fib_levels)

# Print Results 
for level, price in zip(fib_levels, retracement_levels):
    print(f"Fibonnaci {level:.3f} retracement level: ${price:.2f}")
    

```

#### 🔧 **Exercise 7: Advanced Function Combination**

Create a comprehensive trading analysis function using all concepts.

**Function should:**
- Take: `positions` list of (ticker, shares, price) tuples
- Use `map()` to calculate position values
- Use lambda to filter profitable positions
- Return: total_value, profitable_positions, loss_positions

**Test with:** `[("AAPL", 100, 150), ("GOOGL", 50, 2800), ("MSFT", 200, 400)]`


```python
def analyse_positions(positions):
    # Calculate position values using map:
    position_values = list(map(lambda x: x[1] * x[2], positions))

    # Calculate total portfolio value
    total_value = sum(position_values)

    # filter profitable positions ( assuming prices are 10% higher)

    profitable_positions = list(filter(lambda pos: (pos[1] * (pos[2] * 1.1)) > (pos[1] * pos[2]), positions))

    # filter profitable positions ( assuming prices are 10% higher)
    loss_positions = list(filter(lambda pos: (pos[1] * (pos[2]*1.1)) <= (pos[1] * pos[2]), positions))
    return total_value, profitable_positions, loss_positions

# Test the function
test_positions = [("AAPL", 100, 150), ("GOOGL", 50, 2800), ("MSFT", 200, 400)]
total, profitable, losses = analyse_positions(test_positions)

print(f"Total Portfolio Value: ${total:,.2f}")
print("\nProfitable Positions:")
for pos in profitable:
    print(f"{pos[0]}: {pos[1]} shares at {pos[2]:.2f}")
print("\nLoss Positions:")
for pos in losses:
    print(f"{pos[0]}: {pos[1]} shares at {pos[2]:.2f}")
```
