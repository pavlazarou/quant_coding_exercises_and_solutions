#### 📊 **Exercise 1: Robust Portfolio Calculator**

 Create a portfolio return calculator with two key functions:

 1. **validate_inputs(weights, returns)**: 
    - Validates that weights and returns are lists
    - Checks lists have equal length
    - Verifies all values can convert to float
    - Ensures weights sum to 1
    - Returns (is_valid, error_message) tuple

 2. **calculate_portfolio_return(weights, returns)**:
    - Uses validate_inputs() to check inputs
    - Calculates weighted sum of returns
    - Handles calculation errors
    - Returns (return_value, error_message) tuple

Test with both valid inputs (e.g. [0.4, 0.3, 0.3], [0.1, 0.05, 0.15]) 
 and invalid inputs (e.g. [0.5, 0.7], [0.1, 0.2])


```python
def validate_inputs(weights, returns):

    # check if inputs are lists
    if not isinstance(weights, list) or not isinstance(returns, list):
        return False, "Weights and returns must be lists"
    
    # Check Lengths Match
    if len(weights) != len(returns):
        return False, "Weights and returns must have same length"
    
    # Check values can convert to floats
    try:
        weights = [float(w) for w in weights]
        returns = [float(r) for r in returns]
    except (ValueError, TypeError):
        return False, "All values must be numeric"
    
    # Check if sum of weights equal 1
    sum_weights = sum(weights)
    if abs(sum_weights - 1) > 0.0001:
        return False, "Weights must sum to 1"
    return True,""

def calculate_portfolio_returns(weights, returns):

    # Validate inputs
    is_valid, error_msg = validate_inputs(weights, returns)
    if not is_valid:
        return None, error_msg
    try:
        # Calculate weighted sum
        portfolio_return = 0
        for w, r in zip(weights, returns):
            portfolio_return += w * r
        return portfolio_return, ""
    except Exception as e:
        return None, f"Calculation error: {str(e)}"
    
# valid Case
weights = [0.4, 0.3, 0.3]
returns = [0.1, 0.05, 0.15]
port_return, error = calculate_portfolio_returns(weights, returns)

if error:
    print(f"Error: {error}")
else:
    print(f"Portfolio Return: {port_return}")

# Error Case
weights = [0.5, 0.7]
returns = [0.1, 0.2]
port_return, error = calculate_portfolio_returns(weights, returns)

if error:
    print(f'Error: {error}')
else:
    print(f"Portfolio Return: {port_return}")

    

        

                               
```

#### 🎯 **Exercise 2: Safe Data Access System**

 Build a safe data access system for financial data with the following functions:

 1. Create a `validate_data()` function that:
    - Takes a value, data type, and optional min/max values
    - Returns (True/False, error message)
    - Validates both type and numeric ranges

 2. Create a `safe_get()` function that:
    - Takes a category and key to look up in financial_data
    - Returns (value, error message)
    - Handles missing data with defaults
    - Logs all access attempts

 3. Create a `get_access_stats()` function that:
    - Returns a dict with access statistics
    - Tracks total accesses and error rates

 Handle all errors with try-except blocks.


```python
# Sample financial data store
financial_data = {
    'stocks': {'AAPL': 150.0, 'GOOGL': 2800.0},
    'rates': {'USD': 0.02, 'EUR': 0.01}
}
access_log = []
error_log = []

def validate_data(value, data_type, min_val=None, max_val=None):
    try:
        # Validate Data Type
        if not isinstance(value, data_type):
            raise TypeError(f"Expected { data_type.__name__}, got {type(value).__name__}")

        # Range Check
        if min_val is not None and value < min_val:
            raise ValueError(f" Value {value} below minimum {min_val}")
        if max_val is not None and value > max_val:
            raise ValueError(f"Value { value} above maximum {max_val}")
        
        return True,""
    except Exception as e:
        return False, str(e)
    
def safe_get(category, key, default = None):
    try:
        # Log access
        access_log.append((category, key))

        # Primary Data access
        if category in financial_data and key in financial_data[category]:
            return financial_data[category][key],""
        
        # Fall back to default
        return default, f"Data not found for {category}/{key}, using default"
    except Exception as e:
        error_log.append((category, key, str(e)))
        return f"Error accessing data : {str(e)}"

def get_access_stats():
    return {
        'total_access' : len(access_log),
        'total_errors' : len(error_log),
        'error_rate' : len(error_log) / len(access_log) if access_log else 0
    }

# Example usage
# Valid data access
value, error = safe_get('stocks', 'AAPL')
print(f"AAPL stock price: {value}" if not error else f"Error: {error}")

# Invalid data access
value, error = safe_get('stocks', 'INVALID', default=0.0)
print(f"Invalid stock price: {value}" if not error else f"Error: {error}")

# Data Validation
valid, error = validate_data(100.0, float, min_val=0, max_val=1000)
print(f"Validation {'passed' if valid else 'failed'}: {error}")

# Monitor access patterns
stats = get_access_stats()
print(f"Access Statistics: {stats}")



```

#### 🔧 **Exercise 3: Error Monitoring Dashboard**

 Create an error monitoring system with the following functions:

 1. Implement `track_error(error_type, message)` to:
    - Record errors in a list
    - Print error details
    - Alert when error count exceeds threshold (3) for a type

 2. Create `show_stats()` to display:
    - Total number of errors
    - Count of each error type
    - Print "No errors recorded" if empty

 3. Add `get_recovery(error_type)` that returns actions:
    - 'DataError' -> 'Retry data fetch'
    - 'ValueError' -> 'Validate input'
    - 'ConnectionError' -> 'Reconnect'
    - Default -> 'No recovery action defined'

 Test the system by:
 - Logging multiple ValueError and DataError instances
 - Displaying error statistics
 - Getting recovery actions for different error types


```python
error_list = []
alert_threshold = 3

def track_error(error_type, message):
    error = (error_type, message)
    error_list.append(error)
    count = sum(1 for e in error_list if e[0] == error_type)
    print(f"Error Logged: {error_type} - {message}")

    if count >= alert_threshold:
        print(f"Alert: {error_type} errors exceeded threshold!")

def show_stats():
    if not error_list:
        return print("No errors recorded")
    
    counts = {}
    for type, _ in error_list:
        counts[type] = counts.get(type,0) + 1
    
    print("\nError Analysis:")
    print(f"Total errors: {len(error_list)}")
    print(("Error counts:", counts))

def get_recovery(error_type):
    actions = {
        'DataError' : 'Retry data fetch',
        'ValueError' : 'Validate input',
        'ConnectionError' : 'Reconnect',
    }
    return actions.get(error_type, 'No recovery action defined')

# Example usage
track_error('ValueError', 'Invalid input')
track_error('DataError', 'Missing data') 
track_error('ValueError', 'Bad format')
track_error('ValueError', 'Wrong type')

show_stats()
print("\nRecovery action:", get_recovery('ValueError'))
```
